```sql
%python
from pyspark.sql.functions import *
import pyspark.pandas as pd
```

```sql
%python
display(spark.read.table("country_data.country_ing.continent_ing"))
```

```sql
%python
df = spark.read.table("country_data.country_ing.continent_ing")

#comando usado para mudar index da tabela
df_transposed = spark.createDataFrame(df.toPandas().T.reset_index())
```

```sql
%python
display(df_transposed)
```

```sql
%python
#renomear colunas
df_renamed = df_transposed.withColumnRenamed("index","country_id").withColumnRenamed("0","continent_id")
df_renamed.write.mode("overwrite").saveAsTable("country_data.country_silver.continent_silver")
```
