```sql
%python
from pyspark.sql.functions import *
```

```sql
%python
df_names = spark.read.table("country_data.country_silver.names_silver")
df_continent_silver = spark.read.table("country_data.country_silver.continent_silver")
df_continent_ucs = spark.read.table("country_data.country_ucs.continent_gold")
df_iso3 = spark.read.table("country_data.country_silver.iso3_silver")
df_phone  = spark.read.table("country_data.country_silver.phone_silver")
df_currency = spark.read.table("country_data.country_silver.currency_silver")
```

```sql
%python
df_final_result = (
    df_names
    .join(df_continent_silver, on='country_id', how="left")
    .join(df_continent_ucs, on='continent_id', how="left")
    .join(df_iso3, on='country_id', how="left")
    .join(df_phone, on='country_id', how="left")
    .join(df_currency, on='country_id', how="left")
    .filter(col("continent_id") != "AN")
    .select(
        col("iso3_id").alias("country_id"),
        col("country_name"),
        col("continent_name"),
        col("phone_code"),
        col("currency_name")
    )
)

df_final_result.display()
```

```sql
%python
df_final_result.write.mode("overwrite").saveAsTable("country_data.country_ucs.country_details")
```
