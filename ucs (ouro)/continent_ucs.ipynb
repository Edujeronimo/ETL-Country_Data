```sql
%python
from pyspark.sql.functions import *
```

```sql
%python
df_silver = spark.read.table("country_data.country_silver.continent_silver")
```

```sql
%python
df_distinct = df_silver.select("continent_id").distinct()
```

```sql
%python
from pyspark.sql.functions import when, col

df_gold =(
    df_distinct.withColumn (
        "continent_name",
          when(col("continent_id") == "OC", "Oceania")
         .when(col("continent_id") == "AF", "Africa")
         .when(col("continent_id") == "EU", "Europe")
         .when(col("continent_id") == "NA", "North America")
         .when(col("continent_id") == "SA", "South America")
         .when(col("continent_id") == "AS", "Asia")
         .when(col("continent_id") == "AN", "Antarctic")
         .otherwise("Not found")
    )
)
```

```sql
%python
df_gold.write.mode("overwrite").saveAsTable("country_data.country_ucs.continent_gold")
```
